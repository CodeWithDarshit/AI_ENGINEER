```python
import numpy as np
import pandas as pd
```

```python
student = pd.DataFrame({
  'id':[1,2,3,4,5],
  'name':['ravi', 'ravi2','ravi3','ravi4','ravi5'],
  'class':[12,12,1,3,5]
})

student2 = pd.DataFrame({
   'id':[1,2,3,6,7],
  'name':['ravi', 'ravi2','ravi3','ravi4','ravi5'],
  'class':[12,12,1,3,5]
})
```

```python
student
```

```python
student2
```

```python
pd.merge(student,student2, on="id",how="inner")
```

```python
pd.merge(student,student2,on="id",how="outer")
```

```python
pd.merge(student,student2,on='id', how='left')
```

```python
pd.merge(student,student2, on='id', how='right')
```
